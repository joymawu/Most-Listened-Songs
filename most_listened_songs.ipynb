```python
import pandas as pd
import matplotlib as plt
```

```python
# Load the CSV file using pandas
df = pd.read_csv('/content/most-listened-songs - Sheet1.csv')

# Display the loaded data
display(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style of the plots
sns.set_theme(style="whitegrid")

# Initialize the matplotlib figure
plt.figure(figsize=(10, 6))

# Create a barplot sorted by Plays with explicit hue to avoid FutureWarnings
sns.barplot(
    x="Plays",
    y="Title",
    data=df.sort_values("Plays", ascending=False),
    hue="Title",
    palette="viridis",
    legend=False
)

# Add titles and labels
plt.title("Most Listened Songs by Number of Plays", fontsize=16)
plt.xlabel("Number of Plays", fontsize=13)
plt.ylabel("Song Title", fontsize=13)

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count the occurrences of each genre
genre_counts = df['Genre'].value_counts()

# Set up the figure
plt.figure(figsize=(8, 8))

# Choose a pleasant color palette
colors = sns.color_palette('pastel')[0:len(genre_counts)]

# Create the pie chart
plt.pie(
    genre_counts,
    labels=genre_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=colors,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5}
)

# Set title and show the plot
plt.title("Distribution of Music Genres", fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
